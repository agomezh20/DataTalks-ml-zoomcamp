## Question 4

First, we serve the model with the command

```bash
uv run uvicorn predict:app --host 0.0.0.0 --port 9696
```

With the previous question costumer as default.

To enter the new costumer, we execute

```bash
curl -s http://localhost:9696/predict \
  -H 'Content-Type: application/json' \
  -d '{
    "lead_source": "organic_search",
    "industry": "technology",
    "employment_status": "employed",
    "location": "europe",
    "number_of_courses_viewed": 4,
    "annual_income": 80304.0,
    "interaction_count": 7,
    "lead_score": 0.74
  }'
```

This gives us the response **{"conversion_probability":0.769799,"conversion":true}**.
