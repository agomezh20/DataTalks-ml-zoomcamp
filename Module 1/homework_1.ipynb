### Q7. Sum of weights

```python
asian_cars.head(4)
```

```python
asian_cars_weight_year = asian_cars[['vehicle_weight', 'model_year']]
asian_cars_weight_year[0:7]
```

```python
X = asian_cars_weight_year[0:7].to_numpy()
X
```

```python
XTX = np.matmul(X.T, X)
XTX
```

```python
XTX_inverse = np.linalg.inv(XTX)
XTX_inverse
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
y
```

```python
w = np.matmul(XTX_inverse, np.transpose(X))
w = np.matmul(w, y)
w
```

```python
w.sum()
```
